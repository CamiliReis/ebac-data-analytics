# Análise de Desempenho Escolar por Área do Conhecimento

```python
# ============================================
# Cálculo por Área do Conhecimento
# ============================================

# Entrada de dados
portugues = float(input("Nota de Português: "))
matematica = float(input("Nota de Matemática: "))
ciencias = float(input("Nota de Ciências: "))
historia = float(input("Nota de História: "))
sociologia = float(input("Nota de Sociologia: "))
ingles = float(input("Nota de Inglês: "))
educacao_fisica = float(input("Nota de Educação Física: "))
quimica = float(input("Nota de Química: "))
fisica = float(input("Nota de Física: "))
geografia = float(input("Nota de Geografia: "))
biologia = float(input("Nota de Biologia: "))

# --------------------------------------------
# Soma por área do conhecimento
# --------------------------------------------

exatas = matematica + fisica + quimica
humanas = historia + geografia + sociologia
linguagens = portugues + ingles + educacao_fisica
natureza = ciencias + biologia

# --------------------------------------------
# Operações matemáticas
# --------------------------------------------

# Soma total
soma_total = exatas + humanas + linguagens + natureza

# Divisão → média geral
quantidade_disciplinas = 10
media_geral = soma_total / quantidade_disciplinas

# Multiplicação → total ponderado por quantidade de matérias
total_exatas = exatas * 3
total_humanas = humanas * 3
total_linguagens = linguagens * 3
total_natureza = natureza * 2

# --------------------------------------------
# Subtrações → comparação entre TODAS as áreas
# --------------------------------------------

dif_exatas_humanas = exatas - humanas
dif_exatas_linguagens = exatas - linguagens
dif_exatas_natureza = exatas - natureza

dif_humanas_linguagens = humanas - linguagens
dif_humanas_natureza = humanas - natureza

dif_linguagens_natureza = linguagens - natureza

# --------------------------------------------
# Exibição dos resultados
# --------------------------------------------

print("\n--- Soma por Área ---")
print("Exatas:", exatas)
print("Humanas:", humanas)
print("Linguagens:", linguagens)
print("Ciências da Natureza:", natureza)

print("\n--- Comparação entre Áreas ---")
print("Exatas - Humanas:", dif_exatas_humanas)
print("Exatas - Linguagens:", dif_exatas_linguagens)
print("Exatas - Natureza:", dif_exatas_natureza)
print("Humanas - Linguagens:", dif_humanas_linguagens)
print("Humanas - Natureza:", dif_humanas_natureza)
print("Linguagens - Natureza:", dif_linguagens_natureza)

print("\nMédia geral:", media_geral)

# --------------------------------------------
# Condicional → situação do aluno
# --------------------------------------------

if media_geral >= 7:
    print("Situação: Aprovado")
elif media_geral >= 5:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
```
