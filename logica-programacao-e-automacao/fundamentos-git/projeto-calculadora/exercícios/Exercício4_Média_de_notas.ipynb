# Exercicio 4 — Média de notas

```python
nota_portugues= float(input("Qual sua nota em portugues?: "))
print(f"Minha nota em portugues é {nota_portugues}")
print()

nota_matematica= float(input("Qual sua nota em matemática?: "))
print(f"Miha nota em matemática é {nota_matematica}")
print()

media_notas= (nota_portugues + nota_matematica) / 2
print(f"A média das notas é: {media_notas}")
print()

print(media_notas>=6)
```
