# Exercício 3 — Conversor simples

```python
real_brasileiro= float(input("Qual o valor em reais?: "))
print(f"O valor em Reais é: R$ {real_brasileiro:.2f}")

cotacao_dolar= 5

valor_dolar = real_brasileiro / cotacao_dolar
print(f"O valor em dólar é: US$ {valor_dolar:.2f}")
```
